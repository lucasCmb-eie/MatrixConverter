```python
T=17e-6 # Periodo
R=1 # Resistencia
L=1e-3 ## Inductancia

alpha1 = 1/((2*L)/T + R)
beta1 = (R - (2*L)/T)/((2*L)/T + R)

print("alpha1 = ", alpha1)
print("beta1 = ", beta1)

def float_to_q8_24_fmt(value: float) -> str:
    # Escalar
    scaled = int(round(value * (1 << 24)))

    # Ajustar a 32 bits con signo
    if scaled < 0:
        scaled = (1 << 32) + scaled

    # Convertir a binario
    bin_str = format(scaled, '032b')

    # Insertar punto entre Q8 y Q24
    return bin_str[:8] + '.' + bin_str[8:]

def q8_24_to_float_fmt(binary_str: str) -> float:
    # Quitar el punto
    bin_str = binary_str.replace('.', '')
    raw = int(bin_str, 2)

    # Signo (complemento a 2)
    if raw & (1 << 31):
        raw -= 1 << 32

    # Desescalar
    return raw / (1 << 24)

#Convertir los valores de alpha1 y beta1 a binario con formato Q8.24
alpha1_bin = float_to_q8_24_fmt(alpha1)
beta1_bin = float_to_q8_24_fmt(beta1)
print("alpha1_bin = ", alpha1_bin)
print("beta1_bin = ", beta1_bin)

#Convertir los valores de alpha1 y beta1 de binario con formato Q8.24 a float
alpha1_float = q8_24_to_float_fmt(alpha1_bin)
beta1_float = q8_24_to_float_fmt(beta1_bin)
print("alpha1_float = ", alpha1_float)
print("beta1_float = ", beta1_float)
```
